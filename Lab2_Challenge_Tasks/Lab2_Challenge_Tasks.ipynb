```python
# Appendix Exercise 1: Sensor Data Monitoring System
surname = "Antonio"
SEED_NUM = 3

# len("Antonio") is 7, SEED_NUM is 3 -> base factor = 21
base = len(surname) * SEED_NUM
generated_sensor_data = [base + 4, base * 2, base * -10, base + 45, base * 5]

valid_invalid_results = []
classification_results = []

for reading in generated_sensor_data:
    if reading < 0 or reading > 100:
        valid_invalid_results.append("Invalid")
        classification_results.append("Error")
    else:
        valid_invalid_results.append("Valid")
        if reading < 30:
            classification_results.append("Normal")
        elif reading < 60:
            classification_results.append("Warning")
        else:
            classification_results.append("Critical")

print("Assessment Data:")
print("Generated Sensor Data:", generated_sensor_data)
print("Valid/Invalid Results:", valid_invalid_results)
print("Classification Results:", classification_results)
print("Execution Log: Processed readings successfully for", surname, "with seed", SEED_NUM)
print("Final Output: Monitoring complete")
```

```python
# Appendix Exercise 2: Signal Character Analysis
surname = "Antonio"
SEED_NUM = 3

generated_signal = surname + "_SIG_" + str(SEED_NUM * 100)

processed_signal = generated_signal.lower()

character_analysis = []
letter_count = 0
digit_count = 0
special_count = 0

for char in processed_signal:
    if char.isalpha():
        letter_count += 1
        character_analysis.append((char, "Letter"))
    elif char.isdigit():
        digit_count += 1
        character_analysis.append((char, "Digit"))
    else:
        special_count += 1
        character_analysis.append((char, "Special/Symbol"))

if letter_count > 0 and digit_count > 0 and special_count > 0:
    signal_classification = "Complex Alphanumeric Signal (Valid)"
else:
    signal_classification = "Standard Signal"

print("Assessment Data:")
print("Generated Signal:", generated_signal)
print("Processed Signal:", processed_signal)
print("Character Analysis:", character_analysis)
print("Signal Classification:", signal_classification)
print("Execution Log: Processed signal character by character successfully for operator", surname)
print("Final Output: Diagnostic report complete.")
```

```python
# Appendix Exercise 3: Electronic Authentication System
surname = "Antonio"
SEED_NUM = 3

generated_password = surname + "_" + str(SEED_NUM * 111)  # Result: Antonio_333
attempt_limit = SEED_NUM + 2  # Result: 5 allowed attempts

attempts_made = 3  # Successful login within the allowed limit

if attempts_made <= attempt_limit:
    access_result = "Granted"
    final_system_state = "Unlocked"
else:
    access_result = "Denied"
    final_system_state = "Locked Out"

print("Assessment Data:")
print("Generated Password:", generated_password)
print("Attempt Limit:", attempt_limit)
print("Attempts Made:", attempts_made)
print("Access Result:", access_result)
print("Final System State:", final_system_state)
print("Execution Log: Authentication verified successfully for user", surname, "within limit.")
print("Final Output: System access authorized.")
```
